```python
# Input your polytope in your favorite way
def Delta(X, n):
    return Polyhedron(vertices = [[int(i == j) for i in range(n)] for j in X])

def hypergraphic_pol(H):
    n = max(max(X) for X in H)
    return sum(Delta(X, n) for X in H)

def asso(n):
    Ie = []
    for a in range(n):
        for b in range(a, n):
            Ie.append([-binomial(b-a+1, 2)] + [int(a <= i and i <= b) for i in range(n)])
    return Polyhedron(ieqs = Ie, eqns = [[-binomial(n+1, 2)] + [1]*n])

n = 3
P = asso(n)
print(P)
print(P.Hrepresentation())
P.show()

# Check your polytope
Q = P.affine_hull_projection(orthonormal=True, extend = True)
Q.show()

# Save it
Q.plot().save('Downloads/my_polytope_2.stl')
```
